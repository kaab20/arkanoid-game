```python
import os
import requests
from io import BytesIO
import numpy as np
import matplotlib.pyplot as plt
import PIL
```

```python
GITHUB_PATH = 'https://raw.githubusercontent.com/ArielaLevkov/ImageProcessing_TA/main/images/'
```

```python
def read_file(file_name:str, file_path:str=GITHUB_PATH) -> np.ndarray:
    '''
    Accept an image file name (defaulty located in GitHub) and return the image
    in a numpy array.

    Args:
        file_name: the file name
        file_path: the file path.. Default value is Doron's Github repo.

    Returns:
        the image converted into a Numpy array.
    '''
    file_path = os.path.join(file_path, file_name)
    response = requests.get(file_path)
    fp = BytesIO(response.content)

    img_pil = PIL.Image.open(fp)
    return np.array(img_pil, dtype='int16')
```

```python
img = read_file(file_name='house.tiff')
```

```python
def plot_img(img: np.array, figsize:(int,int)=None):
    if figsize:
        plt.figure(figsize = figsize)
    plt.imshow(img, cmap='gray', vmin=0, vmax=255)
```

```python
def calc_cdf(img):
    histogram = np.zeros(256, dtype=np.int64)
    rows, cols = img.shape

    # Count pixels at each gray level
    for r in range(rows):
        for c in range(cols):
            gray = img[r, c]
            histogram[gray] += 1

    # Calculate the normalized cumulative histogram
    cdf = np.zeros(256, dtype=np.float64)
    cumulative = 0
    total_pixels = rows * cols

    for gray in range(256):
        cumulative += histogram[gray]
        cdf[gray] = cumulative / total_pixels

    return cdf


def hist_shape(img_src: np.array, img_dest: np.array) -> np.array:
    '''
    Match the histogram of a source grayscale image
    to the histogram of a reference image.

    Args:
        img_src: Source image in int16 format.
        img_dest: Reference image in int16 format.

    Returns:
        img_new: Adjusted image in int16 format.
    '''

    ######################################################################
    # TODO: Implement image enhancement using histogram Equalization.
    ######################################################################

    # Calculate both cumulative distributions
    cdf_src = calc_cdf(img_src)
    cdf_dest = calc_cdf(img_dest)

    # Build the gray-level mapping
    cv = np.zeros(256, dtype=np.int16)
    j = 0

    for i in range(256):
        while j < 255 and cdf_dest[j] < cdf_src[i]:
            j += 1

        cv[i] = j

    # Create the output image
    rows, cols = img_src.shape
    img_new = np.zeros((rows, cols), dtype=np.int16)

    # Apply the mapping to each pixel
    for r in range(rows):
        for c in range(cols):
            img_new[r, c] = cv[img_src[r, c]]

    ######################################################################
    #                            END OF YOUR CODE                        #
    ######################################################################

    return img_new
```
